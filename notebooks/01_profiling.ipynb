```python
from pathlib import Path
import pandas as pd

# Project root = folder where this notebook lives (graph-music-recs)
project_root = Path("./..").resolve()

# Path to the Kaggle dataset inside data_raw
data_dir = project_root / "data_raw" / "Million Song Dataset + Spotify + Last"

print("Project root:", project_root)
print("Data directory:", data_dir)

# List all files in the dataset folder
print("\nFiles in data_dir:")
for p in sorted(data_dir.iterdir()):
    print(" -", p.name)
```

```python
# Cell 2 – Inspect Music Info.csv

music_path = data_dir / "Music Info.csv"
print("Music Info path:", music_path)

# Load a sample (or full if it's not too large)
music_df = pd.read_csv(music_path)

print("\n=== Music Info: shape ===")
print(music_df.shape)

print("\n=== Music Info: columns ===")
print(music_df.columns.tolist())

print("\n=== Music Info: first 5 rows ===")
display(music_df.head())
```

```python
# Cell 3 – Inspect User Listening History.csv

history_path = data_dir / "User Listening History.csv"
print("User Listening History path:", history_path)

# Load the full file (if it's too big, we can switch to nrows=100000)
history_df = pd.read_csv(history_path)

print("\n=== User Listening History: shape ===")
print(history_df.shape)

print("\n=== User Listening History: columns ===")
print(history_df.columns.tolist())

print("\n=== User Listening History: first 5 rows ===")
display(history_df.head())

# Optional: peek at a few basic stats
print("\n=== Basic value counts (first 5 users & tracks) ===")
if "user_id" in history_df.columns:
    print("Distinct users:", history_df["user_id"].nunique())
    print(history_df["user_id"].value_counts().head())
if "track_id" in history_df.columns:
    print("\nDistinct tracks:", history_df["track_id"].nunique())
    print(history_df["track_id"].value_counts().head())
```

```python
# Cell 4 – Dataset profiling + graph schema planning

# --- Basic stats from both tables ---
total_history_rows = len(history_df)
n_users           = history_df["user_id"].nunique()
n_tracks_history  = history_df["track_id"].nunique()

n_tracks_music    = music_df["track_id"].nunique()
n_artists         = music_df["artist"].nunique()
n_genres_nonnull  = music_df["genre"].notna().sum()

print("=== Summary statistics ===")
print(f"User Listening History rows: {total_history_rows:,}")
print(f"Distinct users:             {n_users:,}")
print(f"Distinct tracks (history):  {n_tracks_history:,}")
print(f"Distinct tracks (music):    {n_tracks_music:,}")
print(f"Distinct artists:           {n_artists:,}")
print(f"Rows with non-null genre:   {n_genres_nonnull:,}")

# --- Track coverage: how many history tracks exist in music info? ---
tracks_history_set = set(history_df["track_id"].unique())
tracks_music_set   = set(music_df["track_id"].unique())

intersection_tracks = len(tracks_history_set & tracks_music_set)
coverage = intersection_tracks / n_tracks_history if n_tracks_history > 0 else 0.0

print("\n=== Track ID coverage ===")
print(f"Tracks in history only:      {n_tracks_history - intersection_tracks:,}")
print(f"Tracks in both tables:       {intersection_tracks:,}")
print(f"Coverage of history tracks by music info: {coverage:.2%}")

# --- Proposed graph schema (for your notes/report) ---
print("\n=== Proposed graph schema ===")
print("Nodes:")
print("  - User(user_id)")
print("  - Song(track_id, name, year, artist, genre, audio features like danceability, energy, etc.)")
print("  - Artist(artist)")
print("  - Genre(genre)  [optional]")
print("\nRelationships:")
print("  - (User)-[:LISTENED {playcount}]->(Song)")
print("  - (Song)-[:PERFORMED_BY]->(Artist)")
print("  - (Song)-[:IN_GENRE]->(Genre)  [optional]")
```

```python
# Cell 5 – Deeper profiling: user activity, track popularity, genres

# --- User activity: how many tracks each user has listened to ---
user_activity = history_df.groupby("user_id")["track_id"].count()

print("=== User activity (tracks per user) ===")
print(user_activity.describe())   # count, mean, std, min, max, quartiles

print("\nTop 5 most active users (by number of tracks listened):")
display(user_activity.sort_values(ascending=False).head(5))

# --- Track popularity: how many users listened to each track ---
track_popularity = history_df.groupby("track_id")["user_id"].count()

print("\n=== Track popularity (users per track) ===")
print(track_popularity.describe())

print("\nTop 5 most popular tracks (by number of users):")
top_tracks = track_popularity.sort_values(ascending=False).head(5).reset_index()
top_tracks = top_tracks.merge(
    music_df[["track_id", "name", "artist"]],
    on="track_id",
    how="left"
)
display(top_tracks)

# --- Genre distribution (optional, for report text) ---
print("\n=== Top 10 genres (non-null) ===")
top_genres = (
    music_df[music_df["genre"].notna()]
    ["genre"].value_counts()
    .head(10)
)
display(top_genres)
```

```python
# Cell 6 – Create a sampled subset for quick experiments

from pathlib import Path

processed_dir = project_root / "data_processed"
sample_dir = processed_dir / "sample"
sample_dir.mkdir(parents=True, exist_ok=True)

print("Saving samples to:", sample_dir)

# 1) Sample a subset of users (e.g., 5,000 users)
n_sample_users = 5000

sample_user_ids = (
    history_df["user_id"]
    .drop_duplicates()
    .sample(n=n_sample_users, random_state=42)
)

history_sample = history_df[history_df["user_id"].isin(sample_user_ids)].copy()

print("Sample history shape:", history_sample.shape)

# 2) Get the tracks that appear in this subset
sample_track_ids = history_sample["track_id"].drop_duplicates()
music_sample = music_df[music_df["track_id"].isin(sample_track_ids)].copy()

print("Sample music info shape:", music_sample.shape)

# 3) Save to CSV for quick testing
history_sample_path = sample_dir / "user_listening_history_sample.csv"
music_sample_path   = sample_dir / "music_info_sample.csv"

history_sample.to_csv(history_sample_path, index=False)
music_sample.to_csv(music_sample_path, index=False)

print("\nSaved:")
print(" -", history_sample_path)
print(" -", music_sample_path)
```

```python
# Cell 7 – Human-readable summary for your report

summary_lines = []

summary_lines.append("=== Dataset Summary (Million Song + Spotify + Last.fm) ===")
summary_lines.append(f"- Total user–track interactions: {total_history_rows:,}")
summary_lines.append(f"- Number of unique users: {n_users:,}")
summary_lines.append(f"- Number of unique tracks (history): {n_tracks_history:,}")
summary_lines.append(f"- Number of unique tracks (music metadata): {n_tracks_music:,}")
summary_lines.append(f"- Number of unique artists: {n_artists:,}")
summary_lines.append(f"- Rows with non-null genre values: {n_genres_nonnull:,}")
summary_lines.append(f"- Coverage of history tracks by music metadata: {coverage:.2%}")

summary_lines.append("")
summary_lines.append("=== Behavioral Patterns ===")
summary_lines.append(f"- Median tracks listened per user: {user_activity.median():.2f}")
summary_lines.append(f"- Mean tracks listened per user: {user_activity.mean():.2f}")
summary_lines.append(f"- Max tracks listened by a single user: {user_activity.max():.0f}")
summary_lines.append(f"- Median users per track: {track_popularity.median():.2f}")
summary_lines.append(f"- Mean users per track: {track_popularity.mean():.2f}")
summary_lines.append(f"- Max users listening to a single track: {track_popularity.max():.0f}")

summary_lines.append("")
summary_lines.append("=== Top Genres (non-null, by count) ===")
for genre, count in top_genres.items():
    summary_lines.append(f"- {genre}: {count}")

print("\n".join(summary_lines))
```
